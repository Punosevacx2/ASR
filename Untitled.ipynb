# Eksperiment: Uticaj kvantizacije i veličine modela na brzinu i tačnost transkripcije srpskog govora

## 1. Uvoz biblioteka

```python
import time
import psutil
import os
import numpy as np
from datasets import load_dataset
from faster_whisper import WhisperModel
from jiwer import wer
```

## 2. Učitavanje podataka

Koristimo Mozilla Common Voice 25.0 srpski test skup. Svaki snimak sadrži MP3 audio fajl i referentni transkript na srpskoj ćirilici.

```python
import pandas as pd

DATA_DIR = os.path.expanduser("~/Documents/master/drugiProjekatZvuk/data/cv-corpus-25.0-2026-03-09/sr")
CLIPS_DIR = os.path.join(DATA_DIR, "clips")

df = pd.read_csv(os.path.join(DATA_DIR, "test.tsv"), sep="\t")
df = df[["path", "sentence"]].dropna().reset_index(drop=True)

print(f"Broj snimaka u test skupu: {len(df)}")
print(df.head())
```

## 3. Odabir uzorka

Za eksperiment koristimo 30 snimaka iz test skupa.

```python
df_sample = df.head(30).copy()

df_sample["full_path"] = df_sample["path"].apply(lambda x: os.path.join(CLIPS_DIR, x))
df_sample = df_sample[df_sample["full_path"].apply(os.path.exists)].reset_index(drop=True)

print(f"Pronađeno snimaka na disku: {len(df_sample)}")
```

## 4. Definisanje kombinacija

Testiramo 6 kombinacija modela i tipa kvantizacije.

```python
kombinacije = [
    ("tiny", "int8"),
    ("tiny", "float32"),
    ("base", "int8"),
    ("base", "float32"),
    ("small", "int8"),
    ("small", "float32"),
]

print("Kombinacije za testiranje:")
for model, kvant in kombinacije:
    print(f"  model={model}, compute_type={kvant}")
```

## 5. Funkcija za testiranje

Za svaku kombinaciju merimo:
- **WER** (Word Error Rate) - stopa greške na nivou reči u poređenju sa referentnim transkriptom
- **RTF** (Real-Time Factor) - odnos između vremena transkripcije i trajanja snimka

```python
def testiraj_kombinaciju(model_size, compute_type, df_sample):
    print(f"\nUčitavam model: {model_size}, compute_type: {compute_type}")
    
    model = WhisperModel(model_size, device="cpu", compute_type=compute_type)
    
    referentni = []
    hipoteze = []
    rtf_lista = []
    
    for _, red in df_sample.iterrows():
        audio_path = red["full_path"]
        referentni_tekst = red["sentence"]
        
        try:
            import av
            container = av.open(audio_path)
            trajanje = container.duration / 1_000_000
            container.close()
        except:
            trajanje = 5.0
        
        pocetak = time.time()
        segmenti, info = model.transcribe(audio_path, language="sr", beam_size=5)
        transkript = " ".join([s.text.strip() for s in segmenti])
        kraj = time.time()
        
        vreme = kraj - pocetak
        rtf = vreme / trajanje if trajanje else None
        
        referentni.append(referentni_tekst)
        hipoteze.append(transkript)
        rtf_lista.append(rtf)
    
    wer_score = wer(referentni, hipoteze)
    srednji_rtf = np.mean([r for r in rtf_lista if r is not None])
    
    return {
        "model": model_size,
        "compute_type": compute_type,
        "WER": round(wer_score, 4),
        "RTF": round(srednji_rtf, 4),
    }
```

## 6. Pokretanje eksperimenta

Pokrećemo transkripciju za sve kombinacije modela i kvantizacije. Ovo će potrajati nekoliko minuta.

```python
rezultati = []

for model_size, compute_type in kombinacije:
    rezultat = testiraj_kombinaciju(model_size, compute_type, df_sample)
    rezultati.append(rezultat)
    print(f"  WER={rezultat['WER']}, RTF={rezultat['RTF']}")

print("\nEksperiment završen!")
```

## 7. Prikaz rezultata

```python
import matplotlib.pyplot as plt

df_rezultati = pd.DataFrame(rezultati)
print(df_rezultati.to_string(index=False))
```

## 8. Vizualizacija

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

labele = [f"{r['model']}\n{r['compute_type']}" for r in rezultati]
wer_vrednosti = [r['WER'] for r in rezultati]
rtf_vrednosti = [r['RTF'] for r in rezultati]

boje = ["#2196F3", "#90CAF9", "#4CAF50", "#A5D6A7", "#FF9800", "#FFCC80"]

ax1.bar(labele, wer_vrednosti, color=boje)
ax1.set_title("Word Error Rate (WER) po kombinaciji")
ax1.set_ylabel("WER")
ax1.set_ylim(0, 1.3)

ax2.bar(labele, rtf_vrednosti, color=boje)
ax2.set_title("Real-Time Factor (RTF) po kombinaciji")
ax2.set_ylabel("RTF")
ax2.axhline(y=1.0, color="red", linestyle="--", label="RTF=1 (realno vreme)")
ax2.legend()

plt.tight_layout()
plt.savefig("rezultati.png", dpi=150)
plt.show()
```
